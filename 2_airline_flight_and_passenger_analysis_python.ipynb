```python
import pandas as pd
import numpy as np

df = pd.read_excel("Airline_Flight_and_Passenger_Analysis_Excel_Cleaned.xlsx")
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df["Passenger_Age"] = df["Passenger_Age"].fillna(df["Passenger_Age"].median())
```

```python
df["Ticket_Fare"].dropna().astype(str).str.contains(r"[₹,$]|,").sum()
```

```python
df["Ticket_Fare"] = (
    df["Ticket_Fare"]
    .astype(str)
    .str.replace(r"[₹$,]", "", regex=True)
    .str.replace(",", "", regex=False)
)
```

```python
df["Ticket_Fare"] = df["Ticket_Fare"].fillna(df["Ticket_Fare"].median())
```

```python
df["Baggage_KG"] = df["Baggage_KG"].fillna(df["Baggage_KG"].median())
```

```python
df["Passenger_Satisfaction"] = df["Passenger_Satisfaction"].fillna(
    df["Passenger_Satisfaction"].median()
)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.dtypes
```

```python
df["Passenger_Age"] = df["Passenger_Age"].astype(int)
df["Passenger_Satisfaction"] = df["Passenger_Satisfaction"].astype(int)
```

```python
df["Delay_Minutes"].astype(str).str.contains("min", case=False).sum()
```

```python
df["Delay_Minutes"] = (
    df["Delay_Minutes"]
    .astype(str)
    .str.replace("min", "", case=False, regex=False)
    .str.strip()
)

df["Delay_Minutes"] = pd.to_numeric(df["Delay_Minutes"], errors="coerce")
```

```python
df.dtypes
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.to_excel("Airline_Flight_and_Passenger_Python_Cleaned.xlsx", index=False)
```

```python
total_passengers = df["Passenger_ID"].nunique()
total_flights = df["Flight_ID"].nunique()
total_revenue = df["Ticket_Fare"].sum()
avg_fare = df["Ticket_Fare"].mean()
avg_delay = df["Delay_Minutes"].mean()

total_passengers, total_flights, total_revenue, avg_fare, avg_delay
```

```python
df["Airline"] = df["Airline"].str.strip().str.title()
```

```python
df["Airline"] = df["Airline"].replace({
    "Indigo": "IndiGo",
    "Spicejet": "SpiceJet"
})
```

```python
airline_analysis = df.groupby("Airline").agg(
    Passengers=("Passenger_ID", "nunique"),
    Flights=("Flight_ID", "nunique"),
    Revenue=("Ticket_Fare", "sum"),
    Avg_Fare=("Ticket_Fare", "mean"),
    Avg_Delay=("Delay_Minutes", "mean")
).sort_values("Passengers", ascending=False)

airline_analysis
```

```python
travel_class_analysis = df.groupby("Travel_Class").agg(
    Passengers=("Passenger_ID", "nunique"),
    Revenue=("Ticket_Fare", "sum"),
    Avg_Fare=("Ticket_Fare", "mean")
).sort_values("Revenue", ascending=False)

travel_class_analysis
```

```python
delay_by_airline = df.groupby("Airline").agg(
    Avg_Delay=("Delay_Minutes", "mean"),
    Max_Delay=("Delay_Minutes", "max")
).sort_values("Avg_Delay", ascending=False)

delay_by_airline
```

```python
cancellation_by_airline = df.groupby("Airline").agg(
    Total_Flights=("Flight_ID", "nunique"),
    Cancelled_Flights=("Cancellation_Status", lambda x: (x == "Cancelled").sum())
)

cancellation_by_airline["Cancellation_Rate_%"] = (
    cancellation_by_airline["Cancelled_Flights"]
    / cancellation_by_airline["Total_Flights"] * 100
).round(2)

cancellation_by_airline.sort_values("Cancellation_Rate_%", ascending=False)
```

```python
flight_status_by_airline = pd.crosstab(
    df["Airline"],
    df["Flight_Status"]
)

flight_status_by_airline
```

```python
route_analysis = df.groupby(
    ["Origin_Airport", "Destination_Airport"]
).agg(
    Passengers=("Passenger_ID", "nunique"),
    Flights=("Flight_ID", "nunique"),
    Revenue=("Ticket_Fare", "sum")
).sort_values("Passengers", ascending=False)

route_analysis.head(10)
```

```python
delay_satisfaction = df.groupby("Passenger_Satisfaction").agg(
    Avg_Delay=("Delay_Minutes", "mean"),
    Passengers=("Passenger_ID", "nunique")
).sort_index()

delay_satisfaction
```
